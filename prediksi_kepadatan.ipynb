```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df = pd.read_excel('jumlah-peserta-didik-sd-2024-sd-mi-sederajat.xlsx')
```

```python
data = df.iloc[3:].copy()
```

```python
data = data.rename(columns={
    'Unnamed: 3': 'Provinsi',
    'Unnamed: 4': 'Negeri',
    'Unnamed: 5': 'Swasta',
    'Unnamed: 6': 'Total'
})
```

```python
data = data[['Provinsi', 'Negeri', 'Swasta', 'Total']]
```

```python
data['Negeri'] = data['Negeri'].astype(str).str.replace('.', '').astype(int)
data['Swasta'] = data['Swasta'].astype(str).str.replace('.', '').astype(int)
data['Total'] = data['Total'].astype(str).str.replace('.', '').astype(int)
```

```python
median_total = data['Total'].median()
data['label'] = data['Total'].apply(lambda x: 1 if x > median_total else 0)
```

```python
X = data[['Negeri','Swasta','Total']]
y = data['label']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)
```

```python
clf = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)
```

```python
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
```

```python
akurasi = accuracy_score(y_test, y_pred)
print(f"Akurasi : {akurasi:.4f}")
```

```python
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
```
